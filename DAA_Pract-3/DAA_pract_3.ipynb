```python
import time

def heapify(arr, n, i):
    # Time Complexity: O(log n) - moves an element down the binary tree
    largest = i
    left, right = 2 * i + 1, 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    # Time Complexity: O(n log n) overall | Space Complexity: O(1) in-place
    n = len(arr)

    # 1. Build max heap: O(n)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # 2. Extract elements: O(n log n)
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]  # Move max to end
        heapify(arr, i, 0)               # Re-heapify

    return arr

# --- Main Program ---
user_input = input("Enter space-separated numbers: ")
data = [int(x) for x in user_input.split()]

print("\nOriginal array:", data)

start = time.perf_counter()
sorted_data = heap_sort(data)
end = time.perf_counter()

print("Sorted array:  ", sorted_data)
print(f"Execution time: {(end - start) * 1000:.4f} ms")
```
