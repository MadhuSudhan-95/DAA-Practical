```python
import time

# 1. Get input from user
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]

# 2. Bubble Sort & measure execution time
start_time = time.perf_counter()

n = len(numbers)
for i in range(n):
    for j in range(0, n - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

end_time = time.perf_counter()

# 3. Print results
print("\nSorted List:", numbers)
print(f"Execution Time: {(end_time - start_time) * 1000:.4f} ms")
print("Time Complexity: Best = O(n), Avg/Worst = O(n²)")
```

```python
import time


def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Merge sorted halves
    result, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


# 1. Get user input
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]

# 2. Sort & measure execution time
start_time = time.perf_counter()
sorted_numbers = merge_sort(numbers)
end_time = time.perf_counter()

# 3. Print results
print("\nSorted List:", sorted_numbers)
print(f"Execution Time: {(end_time - start_time) * 1000:.4f} ms")
print("Time Complexity: Best / Avg / Worst = O(n log n)")
```

```python
import time

# 1. Get user input
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]

# 2. Selection Sort & measure execution time
start_time = time.perf_counter()

n = len(numbers)
for i in range(n):
    min_idx = i
    # Find the smallest element in the unsorted portion
    for j in range(i + 1, n):
        if numbers[j] < numbers[min_idx]:
            min_idx = j
    # Swap the found smallest element with the first unsorted element
    numbers[i], numbers[min_idx] = numbers[min_idx], numbers[i]

end_time = time.perf_counter()

# 3. Print results
print("\nSorted List:", numbers)
print(f"Execution Time: {(end_time - start_time) * 1000:.4f} ms")
print("Time Complexity: Best / Avg / Worst = O(n²)")

```

```python
import time

# 1. Get user input
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]

# 2. Insertion Sort & measure execution time
start_time = time.perf_counter()

for i in range(1, len(numbers)):
    key = numbers[i]
    j = i - 1
    # Move elements greater than key one position ahead
    while j >= 0 and numbers[j] > key:
        numbers[j + 1] = numbers[j]
        j -= 1
    numbers[j + 1] = key

end_time = time.perf_counter()

# 3. Print results
print("\nSorted List:", numbers)
print(f"Execution Time: {(end_time - start_time) * 1000:.4f} ms")
print("Time Complexity: Best = O(n), Avg / Worst = O(n²)")
```

```python
import time


def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


# 1. Get user input
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]

# 2. Quick Sort & measure execution time
start_time = time.perf_counter()
sorted_numbers = quick_sort(numbers)
end_time = time.perf_counter()

# 3. Print results
print("\nSorted List:", sorted_numbers)
print(f"Execution Time: {(end_time - start_time) * 1000:.4f} ms")
print("Time Complexity: Best / Avg = O(n log n), Worst = O(n²)")
```
